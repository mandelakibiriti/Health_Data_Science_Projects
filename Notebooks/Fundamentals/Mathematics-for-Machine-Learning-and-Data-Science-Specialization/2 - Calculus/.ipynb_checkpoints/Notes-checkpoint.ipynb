## Calculus Basics

### The derivative

A function *f* is a quantity that relates two quantities to each other. Let 𝑋 and 𝑌 betwo sets. The subset 𝑓 ⊆ 𝑋 ×𝑌 is a function if for every 𝑥 ∈ 𝑋, there is at most one 𝑦 ∈ 𝑌 such that (𝑥, 𝑦) ∈ 𝑓 which can be indicated with the notation below:
$$
𝑓 ∶ 𝑋 →𝑌
$$

Two different points on a plotted line, any two, call them (𝑥<sub>1</sub>,𝑦<sub>1</sub>), (𝑥<sub>2</sub>,𝑦<sub>2</sub>) and then call the slope of the line:
$$
slope(L) ={\frac{[𝑦_1 - 𝑦_2]}{\text[𝑥_2 - 𝑥_1]}} = \frac{rise}{run}
$$
The difference in the 𝑦's corresponds to a vertical change, while the difference in 𝑥’s corresponds to a horizontal change which defines the *steepness*—the ratio of the change in position (𝑦)  to the change in time(𝑥) —is the *velocity* of that object.

Assume we have a curved function 𝑓(x) let’s approximate steepness using tools we know. Focus on the point labeled A, and call it  A = (𝑥, 𝑓(𝑥)). After a moment of thought, the idea naturally occurs to draw a line between (𝑥, 𝑓(𝑥))  and a nearby point (𝑥′, 𝑓(𝑥′))  have our approximation be the slope of that line.

$$
\text{steepness at point A} \approx {\frac{[𝑓(𝑥′) - 𝑓(𝑥)]}{\text[𝑥′ - 𝑥]}}
$$



Once blessed with the idea of approximating the steepness of at by drawing a line from (𝑥, 𝑓(𝑥)) to some other (𝑥′, 𝑓(𝑥′)) , we neurotically yearn to move 𝑥′ closer to 𝑥. We could move 𝑥′ halfway closer to 𝑥, call this new point $x_1$, and update our slope approximation.

$$
\text{steepness at point A} \approx {\frac{[𝑓(x_1) - 𝑓(𝑥)]}{\text[x_1 - 𝑥]}}
$$

Our yearnings are destined for iteration. Do it again, and again, getting ${\frac{[𝑓(x_2) - 𝑓(𝑥)]}{\text[x_2 - 𝑥]}}$ and ${\frac{[𝑓(x_3) - 𝑓(𝑥)]}{\text[x_3 - 𝑥]}}$, and so on. With each step the line approximation gets better and better, closer and closer to our brain’s intuitive picture of the steepness at A. How do we reason about the “end” of this process? We get a number at every step. If we were to run this loop forever, would these approximate numbers approach some concrete number? If so, we could reasonably call that number the “true” steepness of at A.

That is exactly what *limits* do. A limit is computational machinery that allows one  to say “this sequence of increasingly good approximations would, if followed forever,  end up at a specific value.” The limit of this particular line-approximation-scheme is called the **derivative** - which is the slope at a point.
$$
\text{Lagrange notation}, 𝑓′(𝑥)  = \frac{dy}{dx} = \text{Leibniz notation}, \frac{d}{dx}𝑓(𝑥)
$$

#### Trigonometric functions

The unit circle is a circle with a radius of 1, centered at the origin of a coordinate plane. To understand angle 𝜃 and its relationship with the sine and cosine functions, we use this circle. In the unit circle, the angle 𝜃 is measured starting from the positive x-axis (which is where 𝜃 = 0) and moving counterclockwise around the circle.  

In degrees: A full rotation around the circle is 360<sup>∘</sup>. In radians: A full rotation is 2π radians. This is equivalent to 360<sup>∘</sup>.

Radians are another way of measuring angles, and they’re often preferred in calculus because they have a direct relationship with arc lengths and the unit circle. The key idea is that 1 radian is the angle subtended by an arc of a circle whose length is equal to the radius. The complete rotation around the unit circle corresponds to 2𝜋 radians. Visualizing 2π Radians: 
- When the angle 𝜃 = 0, you are at the point (1, 0) on the unit circle (the positive x-axis). 
- As the angle increases, the point moves counterclockwise along the circle. When 𝜃 = 𝜋 radians (half a full rotation), the point is at (-1, 0) (the negative x-axis). 
- When θ= 2π ​ radians (a quarter of the circle), the point is at (0, -1) (the negative y-axis). 
- A full rotation is 2π radians, bringing you back to (1, 0).






#### Inverse Functions

Let's say that you have a person and the function f does something to the person. It puts a hat on them. What would the function g have to do if it's the inverse? Well, it would just remove the hat so that you get the original person back. In other words, the inverse does the opposite of what f does. It undoes what f does.

Let's say you have the variable x and f puts on a hat on it, so it squares it. Let's see. What would g have to do? Well, undo the square, so remove the hat, g is a function that takes as input x squared and returns x. Before we get to see what f and g are, let's take a look at some notation, g of x and f of x are inverses. We write this as g of x equals f inverse of x with an exponent of minus 1. ***But this is not 1 over f of x. This is purely notation.*** If you take x and apply f and then apply g, you get back the x you started with. Basically, g undoes what f does.



Following the graph below you note that for 𝑓(𝑥) = 𝑥<sup>2</sup> at point (1,1) the corresponding y coordinate is given by g(x) = $\sqrt y$ which is (1,1). Meaning the x,y coordinates in 𝑓(𝑥) = 𝑥<sup>2</sup> are the inverse(squares of the square root) of coordinates in g(x) = $\sqrt y$.

More importantly when determining the derivatives of both 𝑓′(𝑥) and g′(y) you will notice that:

Given:
- $ f(x) = x^2 $
- $ g(y) = \sqrt{y} $

The derivatives are:
- $ f'(x) = 2x $
- $ g'(y) = \frac{1}{2\sqrt{y}} $

The inverse relationship of the derivatives is:
$
g'(y) = \frac{1}{f'(x)} = \frac{1}{2x}
$



### Gradient Descent

Gradient descent is an optimization algorithm used in machine learning to minimize a function (find out how well a function is predicting the target value based on a hypothesis function) by iteratively moving in the direction of the steepest descent as defined by the negative of the gradient. It offers a more systematic way to find the values of w and b, that results in the smallest possible cost, j of w, b.
