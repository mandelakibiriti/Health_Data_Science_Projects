### The derivative

A function *f* is a quantity that relates two quantities to each other. Let 𝑋 and 𝑌 betwo sets. The subset 𝑓 ⊆ 𝑋 ×𝑌 is a function if for every 𝑥 ∈ 𝑋, there is at most one 𝑦 ∈ 𝑌 such that (𝑥, 𝑦) ∈ 𝑓 which can be indicated with the notation below:
$
𝑓 ∶ 𝑋 →𝑌
$

Two different points on a plotted line, any two, call them (𝑥<sub>1</sub>,𝑦<sub>1</sub>), (𝑥<sub>2</sub>,𝑦<sub>2</sub>) and then call the slope of the line:
$
slope(L) ={\frac{[𝑦_1 - 𝑦_2]}{\text[𝑥_2 - 𝑥_1]}} = \frac{rise}{run}
$
The difference in the 𝑦's corresponds to a vertical change, while the difference in 𝑥’s corresponds to a horizontal change which defines the *steepness*—the ratio of the change in position (𝑦)  to the change in time(𝑥) —is the *velocity* of that object.

Assume we have a curved function 𝑓(x) let’s approximate steepness using tools we know. Focus on the point labeled A, and call it  A = (𝑥, 𝑓(𝑥)). After a moment of thought, the idea naturally occurs to draw a line between (𝑥, 𝑓(𝑥))  and a nearby point (𝑥′, 𝑓(𝑥′))  have our approximation be the slope of that line.

$
\text{steepness at point A} \approx {\frac{[𝑓(𝑥′) - 𝑓(𝑥)]}{\text[𝑥′ - 𝑥]}}
$



Once blessed with the idea of approximating the steepness of at by drawing a line from (𝑥, 𝑓(𝑥)) to some other (𝑥′, 𝑓(𝑥′)) , we neurotically yearn to move 𝑥′ closer to 𝑥. We could move 𝑥′ halfway closer to 𝑥, call this new point $x_1$, and update our slope approximation.

$
\text{steepness at point A} \approx {\frac{[𝑓(x_1) - 𝑓(𝑥)]}{\text[x_1 - 𝑥]}}
$

Our yearnings are destined for iteration. Do it again, and again, getting ${\frac{[𝑓(x_2) - 𝑓(𝑥)]}{\text[x_2 - 𝑥]}}$ and ${\frac{[𝑓(x_3) - 𝑓(𝑥)]}{\text[x_3 - 𝑥]}}$, and so on. With each step the line approximation gets better and better, closer and closer to our brain’s intuitive picture of the steepness at A. How do we reason about the “end” of this process? We get a number at every step. If we were to run this loop forever, would these approximate numbers approach some concrete number? If so, we could reasonably call that number the “true” steepness of at A.

That is exactly what *limits* do. A limit is computational machinery that allows one  to say “this sequence of increasingly good approximations would, if followed forever,  end up at a specific value.” The limit of this particular line-approximation-scheme is called the **derivative** - which is the slope at a point.
$
\text{Lagrange notation}, 𝑓′(𝑥)  = \frac{dy}{dx} = \text{Leibniz notation}, \frac{d}{dx}𝑓(𝑥)
$

## Trigonometry Fundamentals

The unit circle is a circle with a radius of 1, centered at the origin of a coordinate plane. To understand angle 𝜃 and its relationship with the sine and cosine functions, we use this circle. In the unit circle, the angle 𝜃 is measured starting from the positive x-axis (which is where 𝜃 = 0) and moving counterclockwise around the circle.  

In degrees: A full rotation around the circle is 360<sup>∘</sup>. In radians: A full rotation is 2π radians. This is equivalent to 360<sup>∘</sup>.

Radians are another way of measuring angles, and they’re often preferred in calculus because they have a direct relationship with arc lengths and the unit circle. The key idea is that 1 radian is the angle subtended by an arc of a circle whose length is equal to the radius. The complete rotation around the unit circle corresponds to 2𝜋 radians. Visualizing 2π Radians: 
- When the angle 𝜃 = 0, you are at the point (1, 0) on the unit circle (the positive x-axis). 
- As the angle increases, the point moves counterclockwise along the circle. When 𝜃 = 𝜋 radians (half a full rotation), the point is at (-1, 0) (the negative x-axis). 
- When θ= $\frac{3}{2π}$ ​ radians ($\frac{3}{4}$ quarter of the circle), the point is at (0, -1) (the negative y-axis). 
- A full rotation is 2π radians, bringing you back to (1, 0).



### Understanding the Relationship Between Arc Length, Angle θ, and π on the Unit Circle

The **unit circle** is a circle with radius 1, centered at the origin. It forms the foundation for understanding trigonometric ratios, radians, and their connection to arc lengths.

#### 1. **Arc Length and Angle (θ) in Radians**

- On any circle, the **arc length** $ s $ subtended by a central angle $ \theta $ (in radians) is:
  $
  s = r\theta
  $
  where $ r $ is the radius.

- On the **unit circle** ($ r = 1 $):
  $
  s = \theta
  $
  This means the arc length **numerically equals the angle in radians**.

#### 2. **Connecting π, Arc Length, and Angles**

- The **full circumference** of the unit circle is $ 2\pi $.
- A **full revolution** is $ 2\pi $ radians, so:
  - Arc length for $ \theta = 2\pi $ radians is $ s = 2\pi $.
  - Arc length for $ \theta = \pi $ radians (half-circle) is $ s = \pi $.
  - Arc length for $ \theta = \frac{\pi}{2} $ radians (quarter-circle) is $ s = \frac{\pi}{2} $.

  - The **arc length** from the positive x-axis to that point is $ \theta $. Since the arc length is the $hypotenus = 1$, the resultant adjacent length and the opposite length will be the coordinates at this point are $ (\cos \theta, \sin \theta) $.

### Comparing the Height and Base of Special Right Triangles

Understanding the side ratios of right triangles with angles 45°, 30°, and 60° is fundamental in trigonometry. Here’s a concise comparison of how to determine the height (opposite side) and base (adjacent side) in each case, assuming the hypotenuse is 1 (unit triangle).

#### 1. 45°-45°-90° Right Triangle

- **Angles:** 45°, 45°, 90°
- **Properties:** The two non-hypotenuse sides are equal.

**Formulas (with hypotenuse = 1):**
- Height (opposite 45°):  $ \sin 45^\circ = \frac{\text{Height}}{1} \implies \text{Height} = \frac{\sqrt{2}}{2} $
- Base (adjacent 45°):   $ \cos 45^\circ = \frac{\text{Base}}{1} \implies \text{Base} = \frac{\sqrt{2}}{2} $

**Result:**  
Both height and base are $ \frac{\sqrt{2}}{2} \approx 0.707 $

#### 2. 30°-60°-90° Right Triangle

- **Angles:** 30°, 60°, 90°
- **Properties:** Sides are in a fixed ratio.

#### For 30° Angle (with hypotenuse = 1):

- Height (opposite 30°):  $ \sin 30^\circ = \frac{\text{Height}}{1} \implies \text{Height} = 0.5 $
- Base (adjacent 30°):   $ \cos 30^\circ = \frac{\text{Base}}{1} \implies \text{Base} = \frac{\sqrt{3}}{2} \approx 0.866 $

#### For 60° Angle (with hypotenuse = 1):

- Height (opposite 60°):  $ \sin 60^\circ = \frac{\text{Height}}{1} \implies \text{Height} = \frac{\sqrt{3}}{2} \approx 0.866 $
- Base (adjacent 60°):   $ \cos 60^\circ = \frac{\text{Base}}{1} \implies \text{Base} = 0.5 $

### Key Takeaways

- In a **45°-45°-90° triangle**, height and base are always equal, each being $ \frac{\sqrt{2}}{2} $ times the hypotenuse.
- In a **30°-60°-90° triangle**, the side opposite 30° is always half the hypotenuse; the side opposite 60° is $ \frac{\sqrt{3}}{2} $ times the hypotenuse.
- These ratios allow you to quickly determine any side when the hypotenuse is known, and are foundational for solving many trigonometric problems.

- **Cosine** gives the x-coordinate (horizontal distance).
- **Tangent** is the slope of the line from the origin to the point ($ \tan \theta = \frac{\sin \theta}{\cos \theta} $).



### Examples

### Example 1: Right Triangle

Suppose you have a right triangle where:
- The angle θ = 30°
- The hypotenuse = 10 units

Find the lengths of the opposite and adjacent sides using trigonometric ratios:

- $ \sin 30^\circ = \frac{\text{Opposite}}{10} = 0.5 \implies \text{Opposite} = 10 \times 0.5 = 5 $
- $ \cos 30^\circ = \frac{\text{Adjacent}}{10} = 0.866 \implies \text{Adjacent} = 10 \times 0.866 = 8.66 $

**Interpretation:**  
For a 30° angle, the side opposite is half the hypotenuse, and the adjacent side is about 86.6% of the hypotenuse.

### Example 2: Slope Using Tangent

If a hill rises 4 meters for every 3 meters of horizontal distance, what is the angle of the hill?

- $ \tan \theta = \frac{\text{Opposite}}{\text{Adjacent}} = \frac{4}{3} $
- $ \theta = \arctan\left(\frac{4}{3}\right) \approx 53.13^\circ $

**Interpretation:**  
The tangent ratio directly tells you the steepness of the hill; a higher ratio means a steeper incline.

### Example 3: Unit Circle Application

For θ = 60° (or $ \frac{\pi}{3} $ radians) on the unit circle:
- $ \cos 60^\circ = 0.5 $ (x-coordinate)
- $ \sin 60^\circ = 0.866 $ (y-coordinate)

**Interpretation:**  
A point at 60° on the unit circle is at (0.5, 0.866). This means if you draw a line at a 60° angle from the origin, it meets the circle at this coordinate.

### Example to Finding the Hypotenuse (x) and Base (y) in a Right Triangle with a 30° Angle

Given a right triangle with one angle $ \theta = 30^\circ $, you can use trigonometric ratios to find the lengths of the hypotenuse ($ x $) and the base ($ y $), provided you know the length of at least one side.

### 1. Understanding the Triangle

- **Angle $ \theta $:** $ 30^\circ $
- **Opposite side:** The side opposite the $ 30^\circ $ angle
- **Adjacent side (base, $ y $):** The side next to the $ 30^\circ $ angle (not the hypotenuse)
- **Hypotenuse ($ x $):** The longest side, opposite the right angle

### 2. Trigonometric Ratios

For angle $ \theta $ in a right triangle:
- $ \sin \theta = \dfrac{\text{Opposite}}{\text{Hypotenuse}} $
- $ \cos \theta = \dfrac{\text{Adjacent (base)}}{\text{Hypotenuse}} $

### 3. Solving for Hypotenuse ($ x $) and Base ($ y $)

#### **Case A: You Know the Opposite Side**

Suppose the length of the side opposite $ 30^\circ $ is $ a $:

- **Hypotenuse ($ x $):**
  $
  \sin 30^\circ = \frac{a}{x} \implies x = \frac{a}{\sin 30^\circ} = \frac{a}{0.5} = 2a
  $

- **Base ($ y $):**
  $
  \cos 30^\circ = \frac{y}{x} \implies y = x \cdot \cos 30^\circ = 2a \cdot 0.866 = 1.732a
  $

#### **Case B: You Know the Base ($ y $)**

Suppose the base (adjacent to $ 30^\circ $) is $ y $:

- **Hypotenuse ($ x $):**
  $
  \cos 30^\circ = \frac{y}{x} \implies x = \frac{y}{\cos 30^\circ} = \frac{y}{0.866}
  $

- **Opposite Side:**
  $
  \sin 30^\circ = \frac{\text{Opposite}}{x} \implies \text{Opposite} = x \cdot \sin 30^\circ = \left(\frac{y}{0.866}\right) \cdot 0.5 = \frac{y}{1.732}
  $

#### **Case C: You Know the Hypotenuse ($ x $)**

Suppose the hypotenuse is $ x $:

- **Opposite Side:**
  $
  \sin 30^\circ = \frac{\text{Opposite}}{x} \implies \text{Opposite} = x \cdot 0.5
  $

- **Base ($ y $):**
  $
  \cos 30^\circ = \frac{y}{x} \implies y = x \cdot 0.866
  $

### 4. Example Calculation

If the hypotenuse $ x $) is 10 units:
- **Base ($ y $):** $ y = 10 \times 0.866 = 8.66 $ units
- **Opposite:** $ 10 \times 0.5 = 5 $ units

|-----------------|----------------------------------|----------------------------|

| Base ($ y $)     | $ x = \frac{y}{0.866} $        | $ y $                    |

### Key Takeaway

- Use $ \sin 30^\circ = 0.5 $ and $ \cos 30^\circ \approx 0.866 $ to relate the sides.
- Knowing any one side, you can use these ratios to find the hypotenuse and base in a right triangle with a $ 30^\circ $ angle.

### Intuition Behind Trigonometric Ratios

Trigonometry is fundamentally about understanding the relationships between angles and side lengths in triangles, especially right triangles. The trigonometric ratios—sine, cosine, and tangent—are ways of expressing these relationships, making it easier to solve problems involving triangles, circles, and periodic phenomena.

### Why Trigonometric Ratios Matter

- **Sine, cosine, and tangent** allow us to relate an angle to the proportions of a triangle’s sides.
- These ratios are constant for a given angle, regardless of the triangle’s size, as long as the triangle is right-angled.
- They form the basis for modeling waves, oscillations, and circular motion in physics and engineering.

#### Inverse Functions

Let's say that you have a person and the function f does something to the person. It puts a hat on them. What would the function g have to do if it's the inverse? Well, it would just remove the hat so that you get the original person back. In other words, the inverse does the opposite of what f does. It undoes what f does.

Let's say you have the variable x and f puts on a hat on it, so it squares it. Let's see. What would g have to do? Well, undo the square, so remove the hat, g is a function that takes as input x squared and returns x. Before we get to see what f and g are, let's take a look at some notation, g of x and f of x are inverses. We write this as g of x equals f inverse of x with an exponent of minus 1. ***But this is not 1 over f of x. This is purely notation.*** If you take x and apply f and then apply g, you get back the x you started with. Basically, g undoes what f does.



Following the graph below you note that for 𝑓(𝑥) = 𝑥<sup>2</sup> at point (1,1) the corresponding y coordinate is given by g(x) = $\sqrt y$ which is (1,1). Meaning the x,y coordinates in 𝑓(𝑥) = 𝑥<sup>2</sup> are the inverse(squares of the square root) of coordinates in g(x) = $\sqrt y$.

More importantly when determining the derivatives of both 𝑓′(𝑥) and g′(y) you will notice that:

Given:
- $ f(x) = x^2 $
- $ g(y) = \sqrt{y} $

The derivatives are:
- $ f'(x) = 2x $
- $ g'(y) = \frac{1}{2\sqrt{y}} $

The inverse relationship of the derivatives is:
$
g'(y) = \frac{1}{f'(x)} = \frac{1}{2x}
