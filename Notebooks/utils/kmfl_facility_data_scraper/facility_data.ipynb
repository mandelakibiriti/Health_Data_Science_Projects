```python
import pandas as pd
import numpy as np
import json
import requests
import time
```

```python
with open('facility.json', 'r') as f:
  data = json.load(f)
```

```python
page_array = []
for i in range(481):
    if i < 481:
        page_array.append(i)
```

```python
mfl_url = "http://api.kmhfr.health.go.ke/api/facilities/material/?access_token=[insert_access_token_here]format=json"
all_facilities = []
```

```python
for i in range(120):
    mfl_get_load = requests.get(mfl_url + f'&page={i}').json()
    time.sleep(10)
    results = mfl_get_load.get("results",[])
    all_facilities.extend(results)
    print(i, len(all_facilities), len(results))
```

```python
len(all_facilities)
```

```python
for i in page_array[120:]:
    mfl_get_load = requests.get(mfl_url + f'&page={i}').json()
    time.sleep(10)
    results = mfl_get_load.get("results",[])
    all_facilities.extend(results)
    print(i, len(all_facilities), len(results))
```

```python
with open('facility_kmfl.json', 'w') as out_file:
    out = json.dumps(all_facilities)
    out_file.write(out)
```
